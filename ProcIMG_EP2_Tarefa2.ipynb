```python
from skimage import io
import matplotlib.pyplot as plt

imagens_dict = {
    'flor_i_d_01': io.imread('./images/flor_i_d_01.jpg'),
    'flor_o_d_01': io.imread('./images/flor_o_d_01.jpg'),
    'flor_i_n_01': io.imread('./images/flor_i_n_01.jpg'),
    'flor_o_n_01': io.imread('./images/flor_o_n_01.jpg')
}

imagens_complemento_dict = {}

for nome, imagem_original in imagens_dict.items():
    imagem_complemento = 255 - imagem_original
    imagens_complemento_dict[nome + "_complemento"] = imagem_complemento

fig, axes = plt.subplots(4, 2, figsize=(10, 20))

for i, (nome, imagem_original) in enumerate(imagens_dict.items()):
    imagem_complemento = imagens_complemento_dict[nome + "_complemento"]
    
    axes[i, 0].imshow(imagem_original)
    axes[i, 0].set_title(f"Original: {nome}")
    axes[i, 0].axis('off')
    
    axes[i, 1].imshow(imagem_complemento)
    axes[i, 1].set_title(f"Complemento: {nome}")
    axes[i, 1].axis('off')
    io.imsave(f'./negative/{nome}_complemento.jpg', imagem_complemento)

plt.tight_layout()
plt.show()
```
